```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

## 결측치 채우기
1. 결측치가 적을 때 삭제하는 방법 - 데이터 자체가 귀하기 때문에 추천X
2. 결측치를 다른 대표값(평균, 중간값)으로 대체하는 방법 

```python
titanic.isnull().sum()
```

### (1) titanic['Embarked'] 결측치 채우기

```python
# Embarked(승성한) 데이터 개수 확인
titanic['Embarked'].value_counts()
```

```python
# Embarked에서 결측치가 있는 데이터를 추출하자
titanic[titanic['Embarked'].isnull()]
```

```python
# Pclass, Embarked
# groupby('기준컬럼')[계산컬럼].계산함수()
# pclass를 기준으로 embarked의 개수를 확인
titanic.groupby('Pclass')['Embarked'].value_counts()
```

```python
# 1등급인 사람들이 'S' 승선항에서 제일 많이 승선
# 결측치 데이터가 1등급인 사람들 -> 결측치를 'S'로 채워주자(fillna())
titanic['Embarked'] = titanic['Embarked'].fillna('S')
```

```python
titanic['Embarked'].isnull().sum()
```

```python
# 데이터 상관관계 확인
# corr() -> 문자형 데이터가 들어가면 에러
titanic.corr(numeric_only = True)
```

```python
# Pclass, Sex, Age를 그룹화해서 Age의 평균값을 구하기(groupby)
# 데이터.groupby('기준컬럼')['계산하고싶은컬럼']
ps_gb = titanic.groupby(['Pclass','Sex'])['Age'].mean().astype('int')
ps_gb
```

```python
# 3클래스의 남성 평균 나이값 가져오기
# data.loc[]
ps_gb.loc[(3,'male')]
```

```python
# 인덱스 6인 데이터 가져오기 -> 가져온 값 3클래스 남성의 평균 나이값 변경
titanic.loc[(6, 'Age')] = ps_gb.loc[(3,'male')]
```

```python
titanic.loc[(6, 'Age')] 
```

```python
titanic[titanic['Age'].isnull()]
```

```python
#인덱스 18인 데이터 가져오기 
titanic.loc[18, 'Age'] = ps_gb.loc[(2,'male')]
# 가져온 값에 ()클래스인 ()의 평균나이값 대입
```

```python
#인덱스 20인 데이터 가져오기 
titanic.loc[20, 'Age'] = ps_gb.loc[(titanic.loc[20,'Pclass'], titanic.loc[20, 'Sex'])]
# 가져온 값에 ()zmffotmaks ()의 평균나이값 대입
```

```python
titanic.loc[20,'Age']
```

```python
# isnull() -> pandas
# isnan() -> numpy
```

```python
# 반복문으로 모든 인덱스에 대해서 결측치가 맞다면 클래스별성별별 평균나이값(ps_gb) 대입
for i in titanic.index : 
    # 성별 컬럼이 결측치인지 조건문을 추가
    if np.isnan(titanic.loc[i, 'Age']) : 
        # 반복해서 작업하고 싶은 결측치 채우는 작업
        titanic.loc[i,'Age'] = ps_gb.loc[(titanic.loc[i,'Pclass'],titanic.loc[i,'Sex'])]
```

```python
#결측치 확인
titanic['Cabin'].isnull().sum()
```

```python
#Cabin컬럼 데이터 확인
titanic['Cabin'].unique()
#라벨값이 너무 많아서 학습에 의미가 있을까?
# 값의 첫글자(Deck)만 남겨놓자
```

## 특성공학 : 새로운 컬럼 생성
- Sibsp 컬럼, Parch 컬럼을 합쳐 Family 컬럼으로 생성

```python
titanic.columns
```

```python
# 함계 탑승한 가족(Sibsp, Parch, 1)을 더한 새로운 컬럼
titanic['Family'] = titanic['SibSp']+titanic['Parch']+1
```

```python
titanic['Family']
```

```python
# Family 와 Survived의 관계 시각화
sns.countplot(titanic, x='Family', hue='Survived')
plt.show()
# 1명 : 사망률이 높다
# 2~4명 : 생존률이 높다.
# 5~11명 : 사망률이 높다.
```

## 비닝(Binning)
- 숫자형 데이터 -> 문자형 데이터
- 1 : Alone, 2 ~ 4 : Small, 5 ~ 11 : Large

```python
# 1. 구간 경계값 설정
bins = [0,1,4,11] 
# 0초과 1이하, 1초과 4이하, 4초과 11이하
# 2. 구간 라벨 설정
labels = ['Alone', 'Small', 'Large']
# 3. 구간화(Cut())
# pd.cut(데이터, bins 구간 경계값 설정(1), Labels = 구간 라벨 설정(2))
titanic['Family'] = pd.cut(titanic['Family'], bins=bins, labels = labels)
```

```python
titanic['Family'].unique()
```

```python
# 업데이트된 Family와 Survived의 관계를 시각화
sns.countplot(titanic, x='Family', hue='Survived')
plt.show()
```

```python
titanic.columns
```

```python
# 불필요한 컬럼 : Name, SibSp, Parch, Ticket 제거
titanic = titanic.drop(['Name', 'SibSp', 'Parch', 'Ticket'], axis=1)
```

```python
# titanic 상관관계 확인
titanic.corr()
# 숫자형 데이터로 변환
# 인코딩
# pd.get_dummise(데이터, dtype = 'int')
titanic_oh = pd.get_dummies(titanic, dtype = 'int')
```

## 데이터 분리

```python
# 문제(x) : Survived를 제외한 나머지 컬럼
# 정답(y) : Survived 컬럼
x = titanic_oh.drop('Survived', axis = 1)
y = titanic_oh['Survived']
```

```python
print(x.shape, y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
# 분리비율 8:2 , 고정규칙 1, 클래스 비율 동일
x_train, x_test, y_train, y_test = train_test_split(
    x,y, #원본데이터(문제, 정답)
    test_size = 0.2, # 나누고 싶은 테스트 데이터의 비율
    random_state = 3, # 랜덤 데이터의 규칙을 고정
    stratify = y # 결과값이 범주형(분류모델)일 때, 훈련/테스트 클래스 비율 동일
)
```

```python
print('훈련 :',x_train.shape, y_train.shape)
print('테스트 :',x_test.shape, y_test.shape)
```

# 5.모델 선택
- 앙상블 모델 중 하나인 베깅, 그 중에서도 랜덤포레스트모델로 학습
- 보팅 : 여러개의 다른 종류의 모델이 예측한 결과들을 투표/평균을 통해 최종 예측하는 모델
- 베깅 : 여러개의 같은 종류의 모델이 에측한 결과들을 투표/평균을 통해 최종 에측하는 모델
- 부스팅 : 여러개의 같은 종류의 모델이 순차적으로 예측을 진행하면서 발생하는 오류 최소화 해나가는 모델

```python
# 모델 불러오기
from sklearn.ensemble import RandomForestClassifier
```

```python
# 모델 객체 생성(하이퍼파라미터 조정)
forest = RandomForestClassifier(n_estimators=200,    #트리의 개수
                               max_features= 5,      #각 트리가 선택할 최대 특성 개수
                               max_depth=4,          #각 트리의 depth단계
                               random_state = 11)    #랜덤시드고정
```

# 6.모델 학습

```python
# 모델학습
forest.fit(x_train, y_train)
```

# 7.모델 평가

```python
# 모델 평가
print('트레인 :',forest.score(x_train,y_train))
print('테스트 :',forest.score(x_test,y_test))
```

```python
# 교차 검증
from sklearn.model_selection import cross_val_score
```

```python
cv_forest = cross_val_score(forest, x_train, y_train, cv = 5)
cv_forest
```

```python
cv_forest.mean().round(2)
```

# 추가 Grid Search 적용해서 최적의 하이퍼 파라미터 찾기 

```python
# 도구 불러오기
from sklearn.model_selection import GridSearchCV
```

```python
# 모델 객체 생성(하이퍼파라미터 조정)
#forest = RandomForestClassifier(n_estimators=200,    #트리의 개수
                               # max_features= 5,      #각 트리가 선택할 최대 특성 개수
                               # max_depth=4,          #각 트리의 depth단계
                               # random_state = 11)    #랜덤시드고정
```

```python
# 1.모델의 파라미터 목록 정의하기
params = {'n_estimators' : range(50, 301, 50),
         'max_features' : range(3, 10, 1),
         'max_depth' : range(3, 10, 1)}
```

```python
# 2. 그리드 서치에 적용할 모델 생성
forest_gs = RandomForestClassifier(random_state=11)
```

```python
# 3. 그리드 서치 객체 생성
grid = GridSearchCV(forest_gs,     #학습할 모델 
                   params,         #하이퍼 파라미터 목록
                   cv=5,           # 내부에서 교차검증 진행
                   n_jobs = -1) # 내 컴퓨터에서 사용 가능한 CPU 코어 모두 사용하여 병렬 연산 진행   
```

```python
# 4. 학습(최적의 하이퍼 파라미터 찾기)
grid.fit(x_train, y_train)
```

```python
# grid가 찾아낸 최적의 파라미터 값 확인 
grid.best_params_
```

```python
# 최적의 하이퍼파라미터 값으로 예측을 수행한 성능값(cv 결과값)
grid.best_score_
```

```python
# 최적의 하이퍼파라미터로 세팅된 모델
forest_best = grid.best_estimator_
```

```python
# 그리드 서치를 통해서 하이퍼파라미터 값을 가진 모델을 통해
# 정확도 86%의 성능을 얻음
# 테스트 데이터 평가
forest_best.score(x_test, y_test)
```
